### Task 2: Analyze Characteristics of Duplicates
1. Identify duplicate rows based on selected columns such as MainBranch, Employment, and RemoteWork. Analyse which columns frequently contain identical values within these duplicate rows.
2. Analyse the characteristics of rows that are duplicates based on a subset of columns, such as MainBranch, Employment, and RemoteWork. Determine which columns frequently have identical values across these rows.
   


```python
## Write your code here
# 1. Identifying the duplicate rows
subset = ["MainBranch", "Employment", "RemoteWork"]
duplicate_groups = df[df.duplicated(subset=subset, keep=False)]
# 2. Analyse which columns frequently contain identical values within these duplicate rows.
duplicate_groups.groupby(subset).size().reset_index(name='Count').sort_values(by='Count', ascending=False)
```

### Task 3: Visualize Duplicates Distribution
1. Create visualizations to show the distribution of duplicates across different categories.
2. Use bar charts or pie charts to represent the distribution of duplicates by Country and Employment.


```python
## Write your code here
# creating bar chart to represent the distribution of duplicates by Country
duplicates_df = df[df.duplicated(keep=False)]
country_duplicates = duplicates_df['Country'].value_counts().sort_values(ascending=False)
plt.figure(figsize=(12, 6))
country_duplicates.head(10).plot(kind='bar', color='skyblue')
plt.title('Top 10 Countries by Number of Duplicate Rows')
plt.xlabel('Country')
plt.ylabel('Number of Duplicate Rows')
plt.xticks(rotation=45, ha='right')
plt.tight_layout() # Adjust layout to make room for labels
plt.show()
```

```python
# creating bar chart to represent the distribution of duplicates by Employment
employment_duplicates = duplicates_df['Employment'].value_counts().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
employment_duplicates.plot(kind='bar', color='salmon')
plt.title('Distribution of Duplicates by Employment Status')
plt.xlabel('Employment Status')
plt.ylabel('Number of Duplicate Rows')
plt.xticks(rotation=45, ha='right')
plt.tight_layout() # Adjust layout to make room for labels
plt.show()
```

```python
plt.figure(figsize=(12, 8)) # Use a slightly larger, more circular figure size
employment_duplicates.plot(kind='pie',
                           autopct='%1.1f%%')
plt.title('Distribution of Duplicates by Employment Status (Pie Chart)')
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

### Task 4: Strategic Removal of Duplicates
1. Decide which columns are critical for defining uniqueness in the dataset.
2. Remove duplicates based on a subset of columns if complete row duplication is not a good criterion.




> Any duplicate ResponseId indicates a true duplicate record that should be removed. I will use this column to define uniqueness.



```python
## Write your code here
unique_column = 'ResponseId'
print(f"Number of rows before removing duplicates: {df.shape[0]}")
df_cleaned = df.drop_duplicates(subset=[unique_column], keep='first')
print(f"Number of rows after removing duplicates: {df_cleaned.shape[0]}")
```
