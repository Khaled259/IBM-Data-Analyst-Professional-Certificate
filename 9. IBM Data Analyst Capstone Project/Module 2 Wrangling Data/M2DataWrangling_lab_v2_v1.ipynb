<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Let's normalize entries in columns Country and EdLevel by mapping the inconsistent values to a consistent format.
import numpy as np

# --- Step 1: Create the Standardization Map for Country Column ---
country_map = {
    # Shorten long official names
    'United Kingdom of Great Britain and Northern Ireland': 'United Kingdom',
    'Iran, Islamic Republic of...': 'Iran',
    'Venezuela, Bolivarian Republic of...': 'Venezuela',
    "Lao People's Democratic Republic": 'Laos',
    'Libyan Arab Jamahiriya': 'Libya',
    'Congo, Republic of the...': 'Republic of the Congo',
    'Micronesia, Federated States of...': 'Micronesia',

    # Consolidate duplicate representations
    'Republic of Korea': 'South Korea',
    "Democratic People's Republic of Korea": 'North Korea',

    # Handle geopolitical entities and simplify names
    'Isle of Man': 'United Kingdom',  # Grouping Crown Dependency with UK
    'Hong Kong (S.A.R.)': 'Hong Kong',
    'Viet Nam': 'Vietnam',
    # np.nan is the standard way to represent missing data in pandas.
    'Nomadic': np.nan
}


# --- Step 2: Apply the Mapping to the 'Country' Column ---
df['Country'] = df['Country'].replace(country_map)
```

```python
# Let's verify the normalization of Country column
display(df['Country'].unique())
```

```python
 #--- Step 1: Create the Standardization Map for EdLevel Column ---
edlevel_map = {
    # Shorten long degree names by removing the parenthetical examples
    'Bachelor’s degree (B.A., B.S., B.Eng., etc.)': 'Bachelor’s degree',
    'Master’s degree (M.A., M.S., M.Eng., MBA, etc.)': 'Master’s degree',
    'Professional degree (JD, MD, Ph.D, Ed.D, etc.)': 'Professional degree',
    'Associate degree (A.A., A.S., etc.)': 'Associate degree',

    # Shorten descriptive categories
    'Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)': 'Secondary school',
    'Some college/university study without earning a degree': 'Some college/university study',

    # Handle the ambiguous 'Something else' category by converting it to a missing value
    # This is a common strategy as it's not a defined educational level.
    'Something else': np.nan,

    # 'Primary/elementary school' is already clean, so no mapping is needed for it.
}



# --- Step 2: Apply the Mapping to the 'Country' Column ---
df['EdLevel'] = df['EdLevel'].replace(edlevel_map)

# --- Step 3: Verify the Mapping to the 'Country' Column ---
display(df['EdLevel'].unique())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Let's identify columns with the highest number of missing values.
missing_values = df.isnull().sum()
missing_values_sorted = missing_values.sort_values(ascending=False)
print(missing_values_sorted)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
# Let's look at the statistic description first to decide to choose the median if the data is skewed
df.describe()
```

```python
## let's impute missing values in numerical columns like ConvertedCompYearly with the mean or median.
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(df['ConvertedCompYearly'].median())
df['ConvertedCompYearly'].isnull().sum()
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Let's impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.
df['RemoteWork'] = df['RemoteWork'].fillna(df['RemoteWork'].mode()[0])
df['RemoteWork'].isnull().sum()
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## let's apply Min-Max Scaling to normalize the `ConvertedCompYearly` column
from sklearn.preprocessing import MinMaxScaler

# Create a MinMaxScaler object
scaler = MinMaxScaler()

# Fit and transform the 'ConvertedCompYearly' column
df['ConvertedCompYearly'] = scaler.fit_transform(df[['ConvertedCompYearly']])


## OR We can use the following code
# df['ConvertedCompYearly'] = (df['ConvertedCompYearly'] - df['ConvertedCompYearly'].min()) / (df['ConvertedCompYearly'].max() - df['ConvertedCompYearly'].min())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Let's Log-transform the ConvertedCompYearly column to reduce skewness
df['ConvertedCompYearly'] = np.log(df['ConvertedCompYearly'])
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
### Firstly let's check `YearsCodePro` column for inconsistencies
display(df['YearsCodePro'].unique())
```

```python
# Convert 'YearsCodePro' to numeric, coercing non-numeric values to NaN
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# Calculate the mean of the numeric 'YearsCodePro' column
mean_years_code_pro = df['YearsCodePro'].mean()

# Fill the missing values (including those that were originally non-numeric) with the mean
df['YearsCodePro'] = df['YearsCodePro'].fillna(mean_years_code_pro)

# Display the first few rows and check for missing values
display(df['YearsCodePro'].head())
print("\nMissing values in 'YearsCodePro' after imputation:")
display(df['YearsCodePro'].isnull().sum())
```

```python
# Let's rescale the values of `YearsCodePro` column to be in range from 0-10
df['Scaled_YearsCodePro'] = (df['YearsCodePro'] - df['YearsCodePro'].min()) / (df['YearsCodePro'].max() - df['YearsCodePro'].min()) * 10
```

```python
## Let's create a new column `ExperienceLevel` based on the `YearsCodePro` column
# The scaling has already been applied in the previous step

# Define the bins and labels for experience levels
bins = [df['Scaled_YearsCodePro'].min() -1 , 2, 5, 10, float('inf')] # Adjust the first bin to include the minimum value
labels = ['Junior', 'Intermediate', 'Senior', 'Expert']

df['ExperienceLevel'] = pd.cut(df['Scaled_YearsCodePro'], bins=bins, labels=labels)

# Display the first few rows of the new column along with relevant original columns
display(df[['ResponseId', 'YearsCodePro', 'Scaled_YearsCodePro', 'ExperienceLevel']].head())
```
