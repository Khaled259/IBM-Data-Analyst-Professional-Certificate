### Step 4: Removing Duplicate Rows


**Task 2: Remove Duplicates**
   1. Remove duplicate rows from the dataset using the drop_duplicates() function.
2. Verify the removal by counting the number of duplicate rows after removal .


````python
# Removing the duplicate rows from the dataset using the drop_duplicates function
df.drop_duplicates(inplace=True)
# Verifying the removal of duplicate rows by counting the duplicates
df.duplicated().sum()
````

### Step 5: Handling Missing Values


**Task 3: Identify and Handle Missing Values**
   1. Identify missing values for all columns in the dataset.
   2. Choose a column with significant missing values (e.g., EdLevel) and impute with the most frequent value.


````python
# Identifying the missing values for all columns in the dataset
df.isna().sum()
````

````python
# Sorting descendingly the number of missing values to find the column of most missing values
df.isna().sum().sort_values(ascending=False).head(15)
````

````python
# The column with significant missing values is "AINextMuch less integrated"
df['AINextMuch less integrated'].value_counts()
````

````python
# Based on the previous output the most frequent value is "Writing code"
df['AINextMuch less integrated'].fillna('Writing code', inplace=True)
# We can use the mode method to replace the missing values with most frequent value (The mode)
#df['AINextMuch less integrated'].fillna(df['AINextMuch less integrated'].mode()[0], inplace=True)
````

````python
# Verifying the replacement of missing values
df['AINextMuch less integrated'].isna().sum()
````

### Step 6: Normalizing Compensation Data


**Task 4: Normalize Compensation Data Using ConvertedCompYearly**
   1. Use the ConvertedCompYearly column for compensation analysis as the normalized annual compensation is already provided.
   2. Check for missing values in ConvertedCompYearly and handle them if necessary.


````python
# Checking for missing values in the target column 'ConvertedCompYearly'
df['ConvertedCompYearly'].isna().sum()
````

````python
# 2. Handling the missing values
#         Since salary data is numerical and often skewed, the MEDIAN is the best choice for imputation.
#              The MEAN can be heavily influenced by a few extremely high salaries.
df['ConvertedCompYearly'].fillna(df['ConvertedCompYearly'].median(), inplace=True)
````

````python
# 3. Display summary statistics to understand the distribution
df['ConvertedCompYearly'].describe()
````



>*The statistical summary of ConvertedCompYearly is a direct consequence of the recent data cleaning step. By replacing all missing salaries with the mode, your code identified 65,000 dolar as the most frequent salary and then filled thousands of empty entries with this single number. This artificially created a massive concentration of data at that exact point, which is why the 25th, 50th (median), and 75th percentiles all snapped to an identical value of $65,000. The mean remains higher at 72,576$$ because it is still being skewed by the original, legitimate high-salary outliers (like the 16.2$$ million max value), but the core distribution of the dataset now reflects the imputation process rather than the organic survey responses.*
