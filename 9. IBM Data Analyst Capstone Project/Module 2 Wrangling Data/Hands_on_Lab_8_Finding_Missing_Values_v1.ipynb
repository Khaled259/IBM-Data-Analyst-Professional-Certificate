## Tasks


<h2>1. Load the Dataset</h2>
<p>
We use the <code>pandas.read_csv()</code> function for reading CSV files. However, in this version of the lab, which operates on JupyterLite, the dataset needs to be downloaded to the interface using the provided code below.

The functions below will download the dataset into your browser:



```python
# Define the URL of the dataset
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"

# Load the dataset into a DataFrame
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### 2. Explore the Dataset
##### Task 1: Display basic information and summary statistics of the dataset.


```python
## Displying basic info about the dataset using info() method
df.info()
```

```python
# Displaying summary statistics of the dataset using descripe() method
df.describe()
```

### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
## Identifying the missing values for all columns using isna() and sum() methods
df.isna().sum()
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
## Visualzing missing values using a heatmap using seaborn library
plt.figure(figsize=(15, 8)) # Increase figure size for better readability
sns.heatmap(df.isna(), cbar=False, yticklabels=False, cmap='viridis')
plt.title('Heatmap of Missing Values in the Dataset')
plt.show()
```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).


```python
## Counting missing rows numbers for specific columns like (CompTotal ,WorkExp)
print('The number of missing rows of CompToatal column is :',df['CompTotal'].isna().sum(),'and the number of missing rows of WorkExp column is :',df['WorkExp'].isna().sum())
```

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
## Identifying the most frequent value in compTotal column
df['CompTotal'].mode()[0]
```

```python
## Identifying the most frequent value in Employment column
df['Employment'].mode()[0]
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
## Imputing missing values in the Employment column with the most frequent value(Employed, full-time)
df['Employment'].fillna(df['Employment'].mode()[0],inplace=True)
```

```python
# Verifying the Imputation process of Employment column
df['Employment'].isna().sum()
```

### 5. Visualizing Imputed Data
##### Task 7: Visualize the distribution of a column after imputation (e.g., `Employment`).


```python
# 1. Geting the INDEX (the names) of the top 20 categories
top_20_categories = df['Employment'].value_counts().head(20).index

# 2. Creating the plot
plt.figure(figsize=(15, 7))

# 3. Using the 'order' parameter to tell Seaborn which categories to plot and in what order
sns.countplot(x='Employment', data=df, order=top_20_categories)

# --- Adding formatting ---
plt.title('Distribution of Top 20 Employment Statuses', fontsize=16)
plt.ylabel('Count')
plt.xlabel('Employment Status')

# Rotating the x-axis labels 45 degrees and align them to the right
plt.xticks(rotation=45, ha='right')

# Adjusting layout to make sure labels don't get cut off
plt.tight_layout()
plt.show()
```
